3. Ask the user to enter a word or short sentence. Display the number of characters, the number of vowels (a, e, i, o, u), and the number of spaces. Use a for loop to examine each character.

```python
text = input("Enter a word or a short text :")
vowel = "aeiouAEIOU"
vowel_count = 0
space_count = 0
for char in text:
    if char in vowel:
       vowel_count += 1
    elif char == ' ':
       space_count += 1
print("Total characters:", len(text))
print("Number of vowels :", vowel_count)
print("Number of Space :", space_count)
```

4. Write a function called calculate_grade(marks) that receives a list of marks. The function should calculate the average and return a grade using these rules: A for average ≥ 80, B for ≥ 70, C for ≥ 60, D for ≥ 50, and F for below 50. Call the function using a list of 5 marks.

```python
def calculate_grade(marks):
  avg = sum(marks)/len(marks)

  if avg >=80:
    return 'A'
  elif avg >=70:
    return 'B'
  elif avg >=60:
    return 'C'
  elif avg >=50:
    return 'D'
  else:
    return 'F'
student_marks = [70, 85,90,96,99]
grade = calculate_grade(student_marks)
print("Final Grade:", grade)
```

5. Create a list of dictionaries representing 4 products. Each dictionary should contain product name, price, and quantity. Use a for loop to calculate the total cost of the cart. Then write a function called apply_discount(total) that returns the final price after a 10% discount when the total is greater than or equal to 2,000; otherwise, no discount is applied.

```python
products = [
    {"name": "Soap", "price": 80, "quantity": 1},
    {"name": "Face Wash", "price": 1500, "quantity": 2},
    {"name": "Cream", "price": 3000, "quantity": 1},
    {"name": "Serum", "price": 5000, "quantity": 1}
]

total = 0

for product in products:
    total += product["price"] * product["quantity"]

print("Total cost:", total)

def apply_discount(total):
    if total >= 2000:
        return total * 0.90
    else:
        return total

final_price = apply_discount(total)

print("Final price after discount:", final_price)
```

7. Using the student records above, display each student’s
name and marks. Count the students who passed (marks ≥ 50), calculate the class average, and identify
the student with the highest mark.

```python
students = [
    {"name": "Rahim", "department": "CSE", "marks": 85, "attendance": 92},
    {"name": "Karim", "department": "CSE", "marks": 48, "attendance": 76},
    {"name": "Sara", "department": "DS", "marks": 91, "attendance": 95},
    {"name": "Nadia", "department": "DS", "marks": 67, "attendance": 84},
    {"name": "Hasan", "department": "CSE", "marks": 73, "attendance": 78}
]

total_marks = 0
passed_count = 0
highest_student = students[0]
print("Student Records")

for student in students:
  print(student['name'], '---', student['marks'])
  total_marks += student['marks']

  if student['marks']>= 50:
    passed_count += 1
  if student['marks'] > highest_student['marks']:
        highest_student = student

average = total_marks / len(students)

print("\nNumber of students passed:", passed_count)
print("Class average:", average)
print("Highest scorer:", highest_student["name"])
print("Highest marks:", highest_student["marks"])
```

8. Calculate each product’s line total using price *
quantity. Calculate the total inventory value and identify the product with the highest line total.


```python
products = [
 {"name": "Notebook", "price": 120, "quantity": 3},
 {"name": "Pen", "price": 20, "quantity": 5},
 {"name": "USB Drive", "price": 850, "quantity": 2},
 {"name": "Mouse", "price": 650, "quantity": 1}
]
total_inventory_cost = 0
highest_product = products[0]

for product in products:
  line_total = product['price'] * product['quantity']
  print(product['name'], "- Line Total:", line_total)

  total_inventory_cost += line_total

  if line_total > highest_product['price'] * highest_product['quantity']:
        highest_product = product

print("\nTotal Inventory Value:", total_inventory_cost)
print("Product with Highest Line Total:", highest_product['name'])
print("Highest Line Total:",
      highest_product['price'] * highest_product['quantity'])
```
