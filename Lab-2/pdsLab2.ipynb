3. Ask the user to enter a word or short sentence. Display the number of characters, the number of vowels (a, e, i, o, u), and the number of spaces. Use a for loop to examine each character.

```python
text = input("Enter a word or a short text :")
vowel = "aeiouAEIOU"
vowel_count = 0
space_count = 0
for char in text:
    if char in vowel:
       vowel_count += 1
    elif char == ' ':
       space_count += 1
print("Total characters:", len(text))
print("Number of vowels :", vowel_count)
print("Number of Space :", space_count)
```

4. Write a function called calculate_grade(marks) that receives a list of marks. The function should calculate the average and return a grade using these rules: A for average ≥ 80, B for ≥ 70, C for ≥ 60, D for ≥ 50, and F for below 50. Call the function using a list of 5 marks.

```python
def calculate_grade(marks):
  avg = sum(marks)/len(marks)

  if avg >=80:
    return 'A'
  elif avg >=70:
    return 'B'
  elif avg >=60:
    return 'C'
  elif avg >=50:
    return 'D'
  else:
    return 'F'
student_marks = [70, 85,90,96,99]
grade = calculate_grade(student_marks)
print("Final Grade:", grade)
```

5. Create a list of dictionaries representing 4 products. Each dictionary should contain product name, price, and quantity. Use a for loop to calculate the total cost of the cart. Then write a function called apply_discount(total) that returns the final price after a 10% discount when the total is greater than or equal to 2,000; otherwise, no discount is applied.

```python
products = [
    {"name": "Soap", "price": 80, "quantity": 1},
    {"name": "Face Wash", "price": 1500, "quantity": 2},
    {"name": "Cream", "price": 3000, "quantity": 1},
    {"name": "Serum", "price": 5000, "quantity": 1}
]

total = 0

for product in products:
    total += product["price"] * product["quantity"]

print("Total cost:", total)

def apply_discount(total):
    if total >= 2000:
        return total * 0.90
    else:
        return total

final_price = apply_discount(total)

print("Final price after discount:", final_price)
```
