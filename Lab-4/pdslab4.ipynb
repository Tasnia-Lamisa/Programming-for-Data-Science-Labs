```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

line plot

```python
x=[1,2,3,4]
y=[10,20,25,10]
plt.plot(x,y)
plt.title("Line Plot")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.show()
```

Scatter plot

```python
plt.scatter([1,2,3],[5,6,7])
plt.title("Scatter plot")
plt.xlabel("X-Axis")
plt.ylabel("Y-Axis")
plt.show()
```

Bar Chart

```python
categories = ["A", "B", "C"]
values = [12, 18, 10]
plt.bar(categories,values)
plt.title("Bar Chart")
plt.xlabel("Categories")
plt.ylabel("Values")
plt.show()
```

Histogram

```python
data = [1, 2, 2, 3, 3, 3, 4]
plt.hist(data)
plt.title("Histogram")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.show()
```

pie chart

```python
values = [40, 35, 25]
labels = ["A", "B", "C"]
plt.pie(
values,
labels=labels,
autopct="%1.1f%%"
)
plt.title("Pie Chart")
plt.show()
```

Subplots

```python
fig, ax = plt.subplots(1, 2)
# Histogram
ax[0].hist([1, 2, 2, 3, 3, 3, 4])
ax[0].set_title("Histogram")

# Pie chart
ax[1].pie(
    [40, 35, 25],
    labels=["A", "B", "C"],
    autopct="%1.1f%%"
)
ax[1].set_title("Pie Chart")
plt.tight_layout()
plt.show()
```

Problem 1

```python
# Load the Titanic dataset
df = pd.read_csv("train.csv")
print(df.head())
df.info()

# Identify numerical and categorical columns
numerical_cols = df.select_dtypes(include=['int64', 'float64']).columns.tolist()
categorical_cols = df.select_dtypes(include=['object']).columns.tolist()

print("\nNumerical Attributes:", numerical_cols)
print("Categorical Attributes:", categorical_cols)
```

problem 2

```python
# Calculate average fare for each passenger class
avg_fare_by_class = df.groupby("Pclass")["Fare"].mean()

# Create line plot
plt.figure(figsize=(8, 5))
plt.plot(avg_fare_by_class.index, avg_fare_by_class.values, marker='o', linestyle='-', color='b')

plt.title("Average Fare Paid by Passenger Class")
plt.xlabel("Passenger Class (Pclass)")
plt.ylabel("Average Fare")
plt.xticks([1, 2, 3])
plt.grid(True)
plt.show()
```

Problem 3

```python
plt.scatter(df["Age"], df["Fare"], alpha=0.6, color='blue')

plt.title("Passenger Age vs. Fare Paid")
plt.xlabel("Age (Years)")
plt.ylabel("Fare")
plt.grid(True)
plt.show()
```

Problem 4

```python
gender_counts = df["Sex"].value_counts()
plt.bar(gender_counts.index, gender_counts.values, color=['blue', 'pink'])

plt.title("Number of Male and Female Passengers")
plt.xlabel("Gender")
plt.ylabel("Number of Passengers")
plt.show()
```

Problem 5

```python
ages = df["Age"].dropna()
plt.hist(ages, bins=20, color='skyblue', edgecolor='black')

plt.title("Distribution of Passenger Ages")
plt.xlabel("Age (Years)")
plt.ylabel("Frequency")
plt.show()
```

Problem 6

```python
survival_counts = df["Survived"].value_counts()
labels = ["Did Not Survive", "Survived"]
plt.pie(
    survival_counts.values,
    labels=labels,
    autopct="%1.1f%%",
    startangle=90,
    colors=['lightcoral', 'lightgreen']
)

plt.title("Proportion of Passenger Survival")
plt.show()
```

Problem 7

```python
fig, ax = plt.subplots(1, 2, figsize=(12, 5))

# Subplot 1: Histogram of passenger ages
ax[0].hist(df["Age"].dropna(), bins=20, color='cornflowerblue', edgecolor='black')
ax[0].set_title("Distribution of Passenger Ages")
ax[0].set_xlabel("Age")
ax[0].set_ylabel("Frequency")

# Subplot 2: Bar chart of passengers per class
pclass_counts = df["Pclass"].value_counts().sort_index()
ax[1].bar(pclass_counts.index.astype(str), pclass_counts.values, color='mediumseagreen')
ax[1].set_title("Number of Passengers by Class")
ax[1].set_xlabel("Passenger Class")
ax[1].set_ylabel("Number of Passengers")

plt.tight_layout()
plt.show()
```

```python
fig, ax = plt.subplots(1, 3, figsize=(16, 5))

# Subplot 1: Number of passengers by gender (Bar Chart)
gender_counts = df["Sex"].value_counts()
ax[0].bar(gender_counts.index, gender_counts.values, color=['steelblue', 'salmon'])
ax[0].set_title("Passengers by Gender")
ax[0].set_xlabel("Gender")
ax[0].set_ylabel("Count")

# Subplot 2: Distribution of passenger fares (Histogram)
ax[1].hist(df["Fare"].dropna(), bins=25, color='orange', edgecolor='black')
ax[1].set_title("Distribution of Passenger Fares")
ax[1].set_xlabel("Fare")
ax[1].set_ylabel("Frequency")

# Subplot 3: Survival proportion (Pie Chart)
survival_counts = df["Survived"].value_counts()
ax[2].pie(
    survival_counts.values,
    labels=["Did Not Survive", "Survived"],
    autopct="%1.1f%%",
    startangle=90,
    colors=['lightcoral', 'lightgreen']
)
ax[2].set_title("Survival Proportion")

plt.tight_layout()
plt.show()
```
