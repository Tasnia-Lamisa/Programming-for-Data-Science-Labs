```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# first 10 rows
df.head(10)
```

```python
# Passengers belonging to first class
print(df[df["Pclass"]==1])
```

```python
# Passenger younger than 18
print(df[df["Age"]<18])
```

```python
# Male passengers who survived
print(df[
    (df["Sex"] == "male") & (df["Survived"]==1)
    ])
```

```python
# Passengers who paid a fair greater than 100
print(df[df["Fare"] > 100])
```

```python
# 10 passengers who paid the highest fare
print(df.sort_values(by="Fare", ascending=False).head(10))
```

```python
# Average passenger age
print(df["Age"].mean())
```

```python
#Count the number of male and female passengers
print(df["Sex"].value_counts())
```

```python
# Average fare of each passenger class
print(df.groupby("Pclass")["Fare"].mean())
```

```python
#Survive rate of each passenger class
print(df.groupby("Pclass")["Survived"].mean()*100)
```

```python
#1
total_passengers = len(df)
print("Total number of passengers:", total_passengers)
```

```python
avg_age = df["Age"].mean()
print("Average age of passenger is:", avg_age)
```

```python
pclass_counts = df["Pclass"].value_counts()
print("Passenger counts by class:",pclass_counts)
print("Class with the highest number:",pclass_counts.idxmax())
```

```python
avg_fare_gender = df.groupby("Sex")["Fare"].mean()
print("Average fare by gender:", avg_fare_gender)
```

```python
survival_gender = df.groupby("Sex")["Survived"].mean()
print("Survival rate by gender:", survival_gender)
print("Gender with higher survival rate:", survival_gender.idxmax())
```

```python
survival_class = df.groupby("Pclass")["Survived"].mean()
print("Survival rate by passenger class:",survival_class)
```
